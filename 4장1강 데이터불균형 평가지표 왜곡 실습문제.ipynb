```python
# TODO
# 기본 모델 / class_weight 모델 비교와 threshold 0.3, 0.5, 0.7 비교를 진행하세요.

# 모델 학습 
base_model = LogisticRegression(max_iter=1000).fit(X_train, y_train)
weighted_model = LogisticRegression(class_weight="balanced", max_iter=1000).fit(X_train, y_train)

# 평가 지표 계산 함수
def evaluate(model, threshold=0.5):
    prob = model.predict_proba(X_test)[:, 1]      # 양성(사기) 확률
    pred = (prob >= threshold).astype(int)        # threshold 기준으로 예측
    return {
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred),
        "F1-score": f1_score(y_test, pred),
        "PR-AUC": average_precision_score(y_test, prob),
    }

# 두 모델 지표를 하나의 DataFrame으로 정리
compare_df = pd.DataFrame({
    "Base LR": evaluate(base_model),
    "Weighted LR": evaluate(weighted_model),
}).T
display(compare_df.round(4))

# weighted 모델의 threshold별 precision/recall
weighted_prob = weighted_model.predict_proba(X_test)[:, 1]
rows = []
for t in [0.3, 0.5, 0.7]:
    pred = (weighted_prob >= t).astype(int)
    rows.append({
        "Threshold": t,
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred),
    })
display(pd.DataFrame(rows).set_index("Threshold").round(4))
```

**답변:**  
Threshold를 낮출때 장점은 놓치는 사기(FN)가 줄어 Recall이 올라가는 것이고, 단점은 정상 거래를 사기로 오탐(FP)하는 일이 늘어 Precision이 떨어지고 추가 검토 비용이 커지는 것이다.
